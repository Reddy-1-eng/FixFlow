## Step 2: Authenticate & Download 80k Dataset (Kaggle Input or Hugging Face Hub)


```python
import os
import json
from pathlib import Path
from huggingface_hub import login, hf_hub_download
from datasets import Dataset

HF_TOKEN = ""
login(token=HF_TOKEN, add_to_git_credential=True)

# Check if dataset is provided via Kaggle Input Dataset or Hugging Face Hub
kaggle_dataset_dir = Path("/kaggle/input/buildfail-dataset")
if (kaggle_dataset_dir / "train.jsonl").exists():
    train_path = str(kaggle_dataset_dir / "train.jsonl")
    val_path = str(kaggle_dataset_dir / "validation.jsonl")
    print(f"Loaded dataset directly from Kaggle Input: {train_path}")
else:
    DATASET_REPO = "krishna1707/buildfail-dataset"
    print(f"Downloading 80k dataset from Hugging Face: https://huggingface.co/datasets/{DATASET_REPO}...")
    train_path = hf_hub_download(repo_id=DATASET_REPO, filename="train.jsonl", repo_type="dataset", token=HF_TOKEN)
    val_path = hf_hub_download(repo_id=DATASET_REPO, filename="validation.jsonl", repo_type="dataset", token=HF_TOKEN)

PROMPT_TEMPLATE = """<|im_start|>system
You are BuildFail-SLM, an expert AI DevOps and software repair model.
Analyze CI/CD failures accurately, identify root causes, and propose non-destructive unified diffs.
<|im_end|>
<|im_start|>user
{instruction}

{input_text}
<|im_end|>
<|im_start|>assistant
{output_text}
<|im_end|>"""

def prepare_dataset(jsonl_path: str) -> Dataset:
    records = []
    with open(jsonl_path, "r", encoding="utf-8") as f:
        for line in f:
            d = json.loads(line)
            inst = d.get("instruction", "")
            inp = d.get("input", "")
            out = d.get("output", "")
            inp_str = inp if isinstance(inp, str) else json.dumps(inp, indent=2)
            out_str = out if isinstance(out, str) else json.dumps(out, indent=2)
            # Pre-truncate logs > 2500 chars to prevent token overflow
            if len(inp_str) > 2500:
                inp_str = inp_str[:2500] + "\n...[truncated]"
            records.append({"text": PROMPT_TEMPLATE.format(
                instruction=inst, input_text=inp_str, output_text=out_str
            )})
    return Dataset.from_list(records)

train_ds = prepare_dataset(train_path)
val_ds = prepare_dataset(val_path)
print(f"✅ Successfully loaded {len(train_ds):,} training examples & {len(val_ds):,} validation examples!")
```

## Step 3: Load Qwen2.5-Coder-7B-Instruct in 4-bit NF4 with Dual GPU Auto-Mapping


```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

BASE_MODEL_NAME = "Qwen/Qwen2.5-Coder-7B-Instruct"

# 4-bit NormalFloat with FP16 compute for Tesla T4 Tensor Cores
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16
)

tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL_NAME, token=HF_TOKEN, trust_remote_code=True)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.model_max_length = 512

print(f"Loading {BASE_MODEL_NAME} in 4-bit across available GPUs...")
model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL_NAME,
    token=HF_TOKEN,
    quantization_config=bnb_config,
    torch_dtype=torch.float16,
    device_map="auto",
    trust_remote_code=True
)

model = prepare_model_for_kbit_training(model, use_gradient_checkpointing=True)

# Ensure all adapter parameters are Float16
for param in model.parameters():
    if param.dtype == torch.bfloat16:
        param.data = param.data.to(torch.float16)

peft_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"]
)

model = get_peft_model(model, peft_config)
model.print_trainable_parameters()
print("Base model Qwen2.5-Coder-7B-Instruct + LoRA adapter successfully allocated across Kaggle GPUs!")
```

## Step 4: High-Throughput SFT Fine-Tuning Loop (80,000+ Quality Samples)


```python
from trl import SFTTrainer, SFTConfig
import trl.trainer.sft_trainer

# 1. Disable TRL's experimental chunked CE patch for multi-GPU partial objects
trl.trainer.sft_trainer._patch_chunked_ce_lm_head = lambda *args, **kwargs: None

# 2. Robust Trainer subclass: uses standard PyTorch CausalLM cross-entropy loss
class RobustSFTTrainer(SFTTrainer):
    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        outputs = model(**inputs)
        loss = outputs.loss
        if loss is not None and loss.dim() > 0:
            loss = loss.mean()
        return (loss, outputs) if return_outputs else loss

# Tuned for 7B parameter model on Kaggle Dual Tesla T4s (32 GB total VRAM) / P100 / A100
training_args = SFTConfig(
    output_dir="./checkpoints_qwen7b",
    per_device_train_batch_size=2,       # Batch size 2 per GPU prevents OOM with 7B model
    gradient_accumulation_steps=8,      # Effective batch size = 16 (or 32 with Dual GPUs)
    learning_rate=1.5e-4,               # Stable learning rate for 7B parameter LoRA
    num_train_epochs=1,                 # Full epoch across 80,000+ quality samples
    logging_steps=25,
    save_strategy="steps",
    save_steps=500,
    eval_strategy="steps",
    eval_steps=500,
    fp16=False,                         # Pure FP16 compute via bitsandbytes without GradScaler crash
    bf16=False,
    optim="paged_adamw_8bit",
    gradient_checkpointing=True,
    max_grad_norm=0.3,
    warmup_steps=150,
    lr_scheduler_type="cosine",
    dataset_text_field="text",
    max_length=512,
    report_to="none"
)

trainer = RobustSFTTrainer(
    model=model,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    processing_class=tokenizer,
    args=training_args
)

print("=" * 75)
print(f"COMMENCING QWEN2.5-CODER-7B TRAINING ON KAGGLE ({len(train_ds):,} SAMPLES)")
print("=" * 75)
trainer.train()
```

## Step 5: Save & Push Fine-Tuned 7B Adapter to Hugging Face Hub


```python
OUTPUT_MODEL_REPO = "krishna1707/buildfail-slm-qwen7b-adapter"

print(f"Pushing trained 7B adapter to Hugging Face: https://huggingface.co/{OUTPUT_MODEL_REPO}...")
trainer.model.push_to_hub(OUTPUT_MODEL_REPO, token=HF_TOKEN, private=False)
tokenizer.push_to_hub(OUTPUT_MODEL_REPO, token=HF_TOKEN, private=False)
print("🎉 Successfully published BuildFail-SLM Qwen2.5-Coder-7B Adapter to Hugging Face Model Hub!")
```

## Step 6: Test & Verify Model Output


```python
test_prompt = PROMPT_TEMPLATE.format(
    instruction="Diagnose the failure root cause and propose a patch for this Docker CI build:",
    input_text="Step 4/10 : RUN pip install -r requirements.txt\nERROR: Could not find a version that satisfies the requirement torch==2.6.0+cu124\nERROR: No matching distribution found for torch==2.6.0+cu124",
    output_text=""
).strip()

device = "cuda" if torch.cuda.is_available() else "cpu"
inputs = tokenizer(test_prompt, return_tensors="pt").to(device)
with torch.no_grad():
    outputs = model.generate(**inputs, max_new_tokens=150, temperature=0.2, top_p=0.9)
output_text = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
print("\n--- Qwen2.5-Coder-7B Model Output ---\n")
print(output_text)
```
